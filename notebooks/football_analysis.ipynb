: WINS/ DRAWS / LOSSES :

```python
query = """
SELECT
    SUM(CASE WHEN home_score > away_score THEN 1 ELSE 0 END) AS home_wins,
    SUM(CASE WHEN home_score = away_score THEN 1 ELSE 0 END) AS draws,
    SUM(CASE WHEN home_score < away_score THEN 1 ELSE 0 END) AS away_wins
FROM matches;
"""
pd.read_sql_query(query, conn)
```

```python
query= """
SELECT
    League,
    ROUND(AVG(home_score + away_score), 2) AS avg_goals
FROM matches
GROUP BY League
ORDER BY avg_goals DESC;
"""
pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    home_team,
    SUM(home_score) AS home_goals
FROM matches
GROUP BY home_team
ORDER BY home_goals DESC
LIMIT 10;
"""
pd.read_sql_query(query, conn)
```

```python
query = """
SELECT 
    away_team,
    SUM(away_score) AS away_goals
FROM matches
GROUP BY away_team
ORDER BY away_goals DESC
LIMIT 10;
"""
pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    team,
    SUM(goals) AS total_goals
FROM (
    SELECT home_team AS team,
home_score AS goals
    FROM matches
    
    UNION ALL
    
    SELECT away_team AS team,
away_score AS goals
    FROM matches
)
GROUP BY team
ORDER BY total_goals DESC
LIMIT 10;
"""
pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    team,
    SUM(goals_conceded) AS total_conceded
FROM (
    SELECT home_team AS team,
away_score AS goals_conceded
    FROM matches
    
    UNION ALL
    
    SELECT away_team AS team,
home_score AS goals_conceded
    FROM matches
)
GROUP BY team
ORDER BY total_conceded ASC
LIMIT 10;
"""
pd.read_sql_query(query, conn)
```

```python
query = """
SELECT 
    team,
    COUNT(*) AS matches,
    SUM(goals_conceded) AS total_conceded
FROM (
    SELECT home_team AS team,
away_score AS goals_conceded
    FROM matches
    
    UNION ALL
    
    SELECT away_team AS team,
home_score AS goals_conceded
    FROM matches
)
GROUP BY team
HAVING COUNT(*) >= 100
ORDER BY total_conceded ASC
LIMIT 10;
""" 
pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    team,
    COUNT(*) AS matches,
    ROUND(AVG(goals_conceded), 2) AS avg_conceded
FROM (
    SELECT home_team AS team,
away_score AS goals_conceded
    FROM matches

    UNION ALL
    
    SELECT away_team AS team,
home_score AS goals_conceded
    FROM matches
) 
GROUP BY team
HAVING COUNT(*) >= 100
ORDER BY avg_conceded ASC
LIMIT 10;
"""
pd.read_sql_query(query, conn)
```

```python
query = """
SELECT 
    team,
    COUNT(*) AS matches,
    ROUND(AVG(goals_scored), 2) AS avg_scored
FROM (
    SELECT home_team AS team,
home_score AS goals_scored
    FROM matches
    
    UNION ALL
    
    SELECT away_team AS team,
away_score AS goals_scored
    FROM matches
)
GROUP BY team
HAVING COUNT(*) >= 100
ORDER BY avg_scored DESC
LIMIT 10;
"""
pd.read_sql_query(query, conn)
```

: Ομάδες με τις Περισσότερες Νίκες :

```python
query = """
SELECT
    team,
    COUNT(*) AS wins
FROM (
    SELECT home_team AS team
    FROM matches
    WHERE home_score > away_score
    
    UNION ALL
    
    SELECT away_team AS team
    FROM matches
    WHERE away_score > home_score
) AS team_wins
GROUP BY team
ORDER BY wins DESC
LIMIT 10;
"""
pd.read_sql_query(query, conn)
```

: HOME vs AWAY performance :

```python
query = """
SELECT
    team,
    SUM(home_wins) AS home_wins,
    SUM(home_draws) AS home_draws,
    SUM(home_losses) AS home_losses,
    SUM(away_wins) AS away_wins,
    SUM(away_draws) AS away_draws,
    SUM(away_losses) AS away_losses,
    COUNT(*) AS total_matches
FROM (
    SELECT home_team AS team,
        CASE WHEN home_score > away_score
        THEN 1 ELSE 0 
        END AS home_wins,
        CASE WHEN home_score = away_score
        THEN 1 ELSE 0 
        END AS home_draws,
        CASE WHEN home_score < away_score
        THEN 1 ELSE 0
        END AS home_losses,
        0 AS away_wins,
        0 AS away_draws,
        0 AS away_losses
    FROM matches
    
    UNION ALL 
    
    SELECT away_team AS team,
        0 AS home_wins,
        0 AS home_draws,
        0 AS home_losses,
        CASE WHEN away_score > home_score
        THEN 1 ELSE 0
        END AS away_wins,
        CASE WHEN away_score = home_score
        THEN 1 ELSE 0
        END AS away_draws,
        CASE WHEN away_score < home_score
        THEN 1 ELSE 0
        END AS away_losses
    FROM matches
) AS team_performance
GROUP BY team
HAVING COUNT(*) >= 100
ORDER BY total_matches DESC
LIMIT 10;
"""
pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    team,
    SUM(home_wins) AS home_wins,
    SUM(home_draws) AS home_draws,
    SUM(home_losses) AS home_losses,
    SUM(away_wins) AS away_wins,
    SUM(away_draws) AS away_draws,
    SUM(away_losses) AS away_losses,
    COUNT(*) AS total_matches
FROM (
    SELECT home_team AS team,
        League,
        CASE WHEN home_score > away_score
        THEN 1 ELSE 0 
        END AS home_wins,
        CASE WHEN home_score = away_score
        THEN 1 ELSE 0 
        END AS home_draws,
        CASE WHEN home_score < away_score
        THEN 1 ELSE 0
        END AS home_losses,
        0 AS away_wins,
        0 AS away_draws,
        0 AS away_losses
    FROM matches
    
    UNION ALL 
    
    SELECT away_team AS team,
        League,
        0 AS home_wins,
        0 AS home_draws,
        0 AS home_losses,
        CASE WHEN away_score > home_score
        THEN 1 ELSE 0
        END AS away_wins,
        CASE WHEN away_score = home_score
        THEN 1 ELSE 0
        END AS away_draws,
        CASE WHEN away_score < home_score
        THEN 1 ELSE 0
        END AS away_losses
    FROM matches
) AS team_performance
WHERE League = 'Premier-league'
GROUP BY team
HAVING COUNT(*) >= 100
ORDER BY total_matches DESC
LIMIT 10;
"""
pd.read_sql_query(query, conn)
```

: STANDINGS / RANKINGS :

```python
query = """
SELECT
    team,
    COUNT(*) as matches,
    SUM(wins) AS wins,
    SUM(draws) AS draws,
    SUM(losses) AS losses,
    SUM(points) AS points,
    SUM(goals_for) AS goals_for,
    SUM(goals_against) AS goal_against,
    SUM(goals_for) - SUM(goals_against) AS goal_difference
FROM (
    SELECT home_team AS team,
        CASE WHEN home_score > away_score THEN 1 ELSE 0
        END AS wins,
        CASE WHEN home_score = away_score THEN 1 ELSE 0
        END AS draws,
        CASE WHEN home_score < away_score THEN 1 ELSE 0
        END AS losses,
        CASE WHEN home_score > away_score THEN 3
             WHEN home_score = away_score THEN 1
             ELSE 0
        END AS points,
        home_score AS goals_for,
        away_score AS goals_against
    FROM matches
    
    UNION ALL
    
    SELECT away_team AS team,
        CASE WHEN away_score > home_score THEN 1 ELSE 0
        END AS wins,
        CASE WHEN away_score = home_score THEN 1 ELSE 0
        END AS draws,
        CASE WHEN away_score < home_score THEN 1 ELSE 0
        END AS losses,
        CASE WHEN away_score > home_score THEN 3
             WHEN away_score = home_score THEN 1
             ELSE 0
        END AS points,
        away_score AS goals_for,
        home_score AS goals_against
    FROM matches
) AS team_stats
GROUP BY team
HAVING COUNT(*) >= 100
ORDER BY points DESC, goal_difference DESC
LIMIT 20;
"""
pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    season_year,
    League,
    team,
    matches,
    wins,
    draws,
    losses,
    points,
    goals_for,
    goals_against,
    goal_difference
FROM (
    SELECT season_year, League, team,
        COUNT(*) AS matches,
        SUM(wins) AS wins,
        SUM(draws) AS draws,
        SUM(losses) AS losses,
        SUM(points) AS points,
        SUM(goals_for) AS goals_for,
        SUM(goals_against) AS goals_against,
        SUM(goals_for) - SUM(goals_against) AS goal_difference,
        ROW_NUMBER() OVER(PARTITION BY season_year, League
                     ORDER BY SUM(points) DESC, SUM(goals_for) - SUM(goals_against) DESC) AS rank
    FROM (
        SELECT
            season_year,
            League,
            home_team AS team,
            CASE WHEN home_score > away_score THEN 1 ELSE 0
            END AS wins,
            CASE WHEN home_score = away_score THEN 1 ELSE 0
            END AS draws,
            CASE WHEN home_score < away_score THEN 1 ELSE 0
            END AS losses,
            CASE WHEN home_score > away_score THEN 3
                 WHEN home_score = away_score THEN 1
                 ELSE 0
            END AS points,
            home_score AS goals_for,
            away_score AS goals_against
        FROM matches
        
        UNION ALL
        
        SELECT
            season_year,
            League,
            away_team AS team,
            CASE WHEN away_score > home_score THEN 1 ELSE 0
            END AS wins,
            CASE WHEN away_score = home_score THEN 1 ELSE 0
            END AS draws,
            CASE WHEN away_score < home_score THEN 1 ELSE 0
            END AS losses,
            CASE WHEN away_score > home_score THEN 3
                 WHEN away_score = home_score THEN 1
                 ELSE 0
            END AS points,
            away_score AS goals_for,
            home_score AS goals_against
        FROM matches
    ) AS team_season_stats
    GROUP BY season_year, League, team
) AS ranked_teams
WHERE rank = 1
ORDER BY season_year DESC, League
LIMIT 50;  
"""
pd.read_sql_query(query, conn)
```

: Comparison Of Countries :

```python
query = """
SELECT
    country,
    COUNT(*) AS matches,
    SUM(home_score + away_score) AS total_goals,
    ROUND(AVG(home_score + away_score), 2) AS avg_goals,
    SUM( 
        CASE WHEN home_score > away_score THEN 1 ELSE 0
        END
    ) AS home_wins,
    SUM(
        CASE WHEN home_score = away_score THEN 1 ELSE 0
        END
    ) AS draws,
    SUM(
        CASE WHEN home_score < away_score THEN 1 ELSE 0
        END
    ) AS away_wins
FROM matches
GROUP BY country
ORDER BY total_goals DESC;
"""
pd.read_sql_query(query, conn)
```
